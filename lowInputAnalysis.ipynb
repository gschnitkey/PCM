# Low Input Analysis

This program does the first work on low input

```python
fldData = pd.read_feather('/content/drive/MyDrive/data/pcm-2025/fldData.feather')
passData  = pd.read_feather('/content/drive/MyDrive/data/pcm-2025/passData.feather')
```

```python
# Create a pandas DataFrame that sums n, p, k, ai by farm, field, year
summed_df = passData.groupby(['farmId', 'fieldId', 'year'])[['n', 'p', 'k', 'ai']].sum().reset_index()

# Display the resulting DataFrame
summed_df
```

```python
# Merge summed_df with fldData
fldData = pd.merge(fldData,
                   summed_df,
                   on=['farmId', 'fieldId', 'year'],
                   how='left')

# Display the fldData with the new columns
fldData[['farmId', 'fieldId', 'year', 'n', 'p', 'k', 'ai']].head()
```

```python
fldData.info()
```

```python
wrkData = fldData[fldData['mrtnLoc'] == 'Central Illinois'].copy()
wrkData = wrkData[wrkData['crop'] == 'corn'].copy()
wrkData = wrkData[wrkData['covStand'] == 'No Cover Crop'].copy()
wrkData = wrkData[wrkData['covStand'] == 'No Cover Crop'].copy()
wrkData = wrkData[wrkData['sprClass'] == 'HIGH'].copy()
wrkData.head()
```

```python
wrkData['pcmField'].unique()
```

```python
# Calculate the average of 'ai' in wrkData
average_ai = wrkData['ai'].mean()
print(f"Average of 'ai': {average_ai:.2f}")
```

```python
# Calculate the average of 'ai' by year in wrkData
average_ai_by_year = wrkData.groupby('year')['ai'].mean().reset_index()
print("Average of 'ai' by Year:")
print(average_ai_by_year)
```

```python
# Calculate the average of 'nit' by year in wrkData
average_nit_by_year = wrkData.groupby('year')['nit'].mean().reset_index()
print("Average of 'nit' by Year:")
print(average_nit_by_year)
```

```python
# Define bins and labels for nRateCat
bins = [-np.inf, 150, 200, 235, np.inf]
labels = [1, 2, 3, 4]

# Create the nRateCat variable using pd.cut
wrkData['nRateCat'] = pd.cut(wrkData['nit'], bins=bins, labels=labels, right=False)

# Display the counts of each category to verify
print("Counts of nRateCat categories:")
print(wrkData['nRateCat'].value_counts())

```

```python
# Calculate the frequency of nRateCat categories by year
category_frequency_by_year = wrkData.groupby(['year', 'nRateCat']).size().unstack(fill_value=0)

print("Frequency of nRateCat Categories by Year:")
display(category_frequency_by_year)
```

```python
# Calculate the average of 'OFR' by category and year
average_ofr_by_category_year = wrkData.groupby(['year', 'nRateCat'])['olr'].mean().unstack(fill_value=0)

print("Average of OFR by nRateCat Category and Year:")
display(average_ofr_by_category_year)
```

```python
# Calculate the average of 'ai' by nRateCat category
average_ai_by_nRateCat = wrkData.groupby('nRateCat')['ai'].mean().reset_index()

print("Average of 'ai' by nRateCat Category:")
display(average_ai_by_nRateCat)
```

```python
# Calculate the average of 'ai' by nRateCat and year
average_ai_by_nRateCat_year = wrkData.groupby(['year', 'nRateCat'])['ai'].mean().unstack(fill_value=0)

print("Average of 'ai' by nRateCat Category and Year:")
display(average_ai_by_nRateCat_year)
```

```python
# Calculate the frequency of 'ai' by year
ai_frequency_by_year = wrkData.groupby(['year', 'ai']).size().unstack(fill_value=0)

print("Frequency of 'ai' by Year:")
display(ai_frequency_by_year)
```

```python
# Define bins and labels for aiCat
bins_ai = [-np.inf, 3, 8, 12, np.inf]
labels_ai = [1, 2, 3, 4]

# Create the aiCat variable using pd.cut
wrkData['aiCat'] = pd.cut(wrkData['ai'], bins=bins_ai, labels=labels_ai, right=True)

# Display the counts of each category to verify
print("Counts of aiCat categories:")
print(wrkData['aiCat'].value_counts())

# Display the first few rows with the new column
display(wrkData[['ai', 'aiCat']].head())
```

```python
# Calculate the average of 'olr' by aiCat and year
average_olr_by_aiCat_year = wrkData.groupby(['year', 'aiCat'])['olr'].mean().unstack(fill_value=0)

print("Average of OLR by aiCat Category and Year:")
display(average_olr_by_aiCat_year)
```

```python
average_yld_by_aiCat = wrkData.groupby('aiCat')['yld'].mean().reset_index()

print("Average of Yield by aiCat Category:")
display(average_yld_by_aiCat)
```

```python
average_yld_by_aiCat_year = wrkData.groupby(['year', 'aiCat'])['yld'].mean().unstack(fill_value=0)

print("Average of Yield by aiCat Category and Year:")
display(average_yld_by_aiCat_year)
```

```python
average_nit_by_aiCat_year = wrkData.groupby(['year', 'aiCat'])['nit'].mean().unstack(fill_value=0)

print("Average of NIT by aiCat Category and Year:")
display(average_nit_by_aiCat_year)
```

```python
average_n_by_aiCat = wrkData.groupby('aiCat')['n'].mean().reset_index()

print("Average of N Rate by aiCat Category:")
display(average_n_by_aiCat)
```

```python
average_n_by_aiCat_year = wrkData.groupby(['year', 'aiCat'])['n'].mean().unstack(fill_value=0)

print("Average of N Rate by aiCat Category and Year:")
display(average_n_by_aiCat_year)
```

```python
# Calculate the average 'n' for each year
average_n_per_year = wrkData.groupby('year')['n'].transform('mean')

# Create the nAvg variable: 1 if 'n' is less than the yearly average, 0 otherwise
wrkData['nAvg'] = (wrkData['n'] < average_n_per_year).astype(int)

print("First 5 rows of wrkData with the new 'nAvg' variable:")
display(wrkData[['year', 'n', 'nAvg']].head())
```

```python
# Calculate the average 'ai' for each year
average_ai_per_year = wrkData.groupby('year')['ai'].transform('mean')

# Create the aiAvg variable: 1 if 'ai' is less than the yearly average, 0 otherwise
wrkData['aiAvg'] = (wrkData['ai'] < average_ai_per_year).astype(int)

print("First 5 rows of wrkData with the new 'aiAvg' variable:")
display(wrkData[['year', 'ai', 'aiAvg']].head())
```

```python
wrkData['low'] = wrkData['nAvg'] + wrkData['aiAvg']
wrkData['low'].unique()

average_olr_by_low_year = wrkData.groupby(['year', 'low'])['olr'].mean().unstack(fill_value=0)

print("Average of OLR by 'low' Category and Year:")
display(average_olr_by_low_year)
```

```python
wrkData['tilLow'] = wrkData['tilStand'].isin(['No Till', 'Strip Till']).astype(int)

print("First 5 rows of wrkData with the new 'tilLow' variable:")
display(wrkData[['tilStand', 'tilLow']].head())
```

```python
wrkData['lowTil'] = wrkData['nAvg'] + wrkData['aiAvg'] + wrkData['tilLow']
wrkData['lowTil'].unique()

average_olr_by_lowTil_year = wrkData.groupby(['year', 'lowTil'])['olr'].mean().unstack(fill_value=0)

print("Average of OLR by 'lowTil' Category and Year:")
display(average_olr_by_lowTil_year)
```

```python
average_yld_by_lowTil_year = wrkData.groupby(['year', 'lowTil'])['yld'].mean().unstack(fill_value=0)

print("Average of Yield by 'lowTil' Category and Year:")
display(average_yld_by_lowTil_year)
```

```python
fields_by_lowTil_year = wrkData.groupby(['year', 'lowTil'])['fieldId'].nunique().unstack(fill_value=0)

print("Number of Fields by 'lowTil' Category and Year:")
display(fields_by_lowTil_year)
```

```python
average_olr_by_aiAvg_year = wrkData.groupby(['year', 'aiAvg'])['olr'].mean().unstack(fill_value=0)

print("Average of OLR by 'aiAvg' Category and Year:")
display(average_olr_by_aiAvg_year)
```

```python
average_yld_by_aiAvg_year = wrkData.groupby(['year', 'aiAvg'])['yld'].mean().unstack(fill_value=0)

print("Average of Yield by 'aiAvg' Category and Year:")
display(average_yld_by_aiAvg_year)
```

```python
count_observations_by_aiAvg_year = wrkData.groupby(['year', 'aiAvg']).size().unstack(fill_value=0)

print("Count of Observations by 'aiAvg' Category and Year:")
display(count_observations_by_aiAvg_year)
```

```python
average_olr_by_nAvg_year = wrkData.groupby(['year', 'nAvg'])['olr'].mean().unstack(fill_value=0)

print("Average of OLR by 'nAvg' Category and Year:")
display(average_olr_by_nAvg_year)
```

```python
average_yld_by_nAvg_year = wrkData.groupby(['year', 'nAvg'])['yld'].mean().unstack(fill_value=0)

print("Average of Yield by 'nAvg' Category and Year:")
display(average_yld_by_nAvg_year)
```
