# Create Field Data

I used this file to read data that Laura gave me and produce a file that I will use for my analysis.  The data came to me in May 10, 2026.

```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import numpy as np


from google.colab import data_table
data_table.enable_dataframe_formatter()
```

```python
'''
The following file came from Laura. I got this on May 10, 2026. It includes data for the
past years: 2015 to 2025.
'''

fldData = pd.read_excel('/content/drive/MyDrive/data/pcm-2025/2015-2025 PCM by-field - to GS.xlsx')

fldData
```

```python
'''
These are the names of the fields.  I copied this over into an
Excel spreadsheet to rename the files.
'''
colNames = pd.DataFrame(fldData.columns, columns=['Field'])
colNames
```

```python
'''
The following frame gives the original field name (field), my newField name, and whether
to keep it or not.  The "field" was copied from the above cell to make the fieldDataMapping.xlsx

'''
newFieldNames = pd.read_excel('/content/drive/MyDrive/data/pcm-2025/fieldDataNameMapping.xlsx')
newFieldNames
```

```python
'''
The following code renames the columns in the field to my new names

'''
for index, row in newFieldNames.iterrows():
  fldData.rename(columns={row['field']: row['newField']}, inplace=True)

fldData
```

```python
'''
The following code deletes the variables that I do not want to keep

'''

for index, row in newFieldNames.iterrows():
  if row['keep'] == 'No':
    fldData.drop(columns=[row['newField']], inplace=True)

fldData
```

```python
'''
create a new numbering system for farms and fields
'''

farmId = pd.DataFrame(fldData['farmIdO'].unique(), columns=['farmIdO'])
farmId['farmId'] = np.arange(len(farmId))
fldData = pd.merge(fldData, farmId, on='farmIdO', how='left')
fldData

fieldId = pd.DataFrame(fldData['fieldIdO'].unique(), columns=['fieldIdO'])
fieldId['fieldId'] = np.arange(len(fieldId))
fldData = pd.merge(fldData, fieldId, on='fieldIdO', how='left')
fldData

farmId.to_csv('/content/drive/MyDrive/data/pcm-2025/farmId.csv', index=False)
fieldId.to_csv('/content/drive/MyDrive/data/pcm-2025/fieldId.csv', index=False)
```

```python
'''
I used this to download and sort into variables that I want to keep
'''

colNames = pd.DataFrame(fldData.columns, columns=['Field'])
colNames
```

```python
'''
The attached file includes the new order for the field names
'''

newOrder = pd.read_excel('/content/drive/MyDrive/data/pcm-2025/mapOfFieldsToKeep.xlsx')
newOrder
```

```python
newOrder = newOrder[newOrder['Num'].notna()]

iFirst = True
for index, row in newOrder.iterrows():
  if iFirst:
    print(row['Field'])
    fldDataC = pd.DataFrame(fldData[row['Field']],columns = [row['Field']])
    iFirst = False
  else:
    print(row['Field'])
    fldDataC[row['Field']] = fldData[row['Field']]

fldDataC
```

```python
fldDataC['tenure'] = fldDataC['tenure'].astype(str)
fldDataC.to_feather('/content/drive/MyDrive/data/pcm-2025/fldData.feather')
fldDataC.to_csv('/content/drive/MyDrive/data/pcm-2025/fldData.csv',index=False)
```
